```python
from PyPDF2 import PdfReader
import re
import nltk
nltk.download("stopwords")
```

```python
reader = PdfReader("Shopify_App_SEO_and_Marketing_Research.pdf")
n = len(reader.pages)
str_lst = []
for i in range(0, n):
    page = reader.pages[i]
    str_lst.append(page.extract_text())

len(str_lst)
```

```python
merged = "\n".join(str_lst).lower().strip()
clean_str = re.sub(r"[^\w\s]", "", merged)
cleaned_str = re.sub(r"\d", "", clean_str)
token = cleaned_str.split(" ")
len(token)
```

```python
word_freq = {}
for word in token:
    if word in word_freq:
        word_freq[word]+=1
    else:
        word_freq[word] = 1

word_freq
```

```python
stop_word = set(stopwords.words("english"))
filtered_token = [token for word in token if word not in stop_word]
print(filtered_token)
```
